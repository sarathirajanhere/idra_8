# E-commerce Sales Analysis

## Objective
Create a complete data analysis notebook using **Python and Pandas** for the provided e-commerce sales dataset.

### Analysis covered
- Loading the CSV dataset into a Pandas DataFrame
- Inspecting the dataset
- Selecting columns and records
- Filtering orders
- Sorting sales records
- Using `groupby()` and aggregation functions
- Analyzing sales by category, city, product, region, and payment method
- Calculating total sales, average sales, highest and lowest sales
- Calculating total quantity sold and number of orders
- Identifying top-performing products and categories
- Writing meaningful observations based on the findings


## 1. Import Pandas

```python
import pandas as pd
import numpy as np

print("Pandas imported successfully.")
```

## 2. Load the Dataset

```python
# Upload Day8_Ecommerce_Sales_Dataset.csv to Google Colab before running this cell.
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

## 3. Inspect the Dataset

```python
print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

## 4. Basic Dataset Information

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nDataset information:")
df.info()
```

## 5. Check for Missing Values

```python
missing_values = df.isnull().sum()
print(missing_values)
```

## 6. Select Specific Columns

```python
# Selecting individual columns
print("Product and Total Sales:")
print(df[["Product", "Total_Sales"]].head(10))

# Selecting multiple useful order details
print("\nSelected order details:")
print(df[["Order_ID", "City", "Category", "Quantity", "Total_Sales"]].head(10))
```

## 7. Filter Records

```python
# Orders with sales greater than 10,000
high_sales = df[df["Total_Sales"] > 10000]
print("Orders with Total_Sales > 10000:")
print(high_sales[["Order_ID", "Customer", "Product", "Total_Sales"]])

# Electronics orders
electronics = df[df["Category"] == "Electronics"]
print("\nNumber of Electronics orders:", len(electronics))

# UPI orders
upi = df[df["Payment_Method"] == "UPI"]
print("Number of UPI orders:", len(upi))
```

## 8. Apply Multiple Filters

```python
# Sports orders with sales greater than 5,000
filtered_orders = df[
    (df["Category"] == "Sports") &
    (df["Total_Sales"] > 5000)
]

print(filtered_orders[
    ["Order_ID", "City", "Product", "Quantity", "Total_Sales"]
])
```

## 9. Sort Records by Sales

```python
# Highest-value orders
top_orders = df.sort_values("Total_Sales", ascending=False).head(10)

print("Top 10 orders by Total Sales:")
print(top_orders[
    ["Order_ID", "Customer", "City", "Product", "Quantity", "Total_Sales"]
])

# Lowest-value orders
print("\nLowest 5 orders by Total Sales:")
print(df.sort_values("Total_Sales").head(5)[
    ["Order_ID", "Customer", "Product", "Total_Sales"]
])
```

## 10. Overall Sales Statistics

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Average Sales per Order: ₹{average_sales:,.2f}")
print(f"Highest Sale: ₹{highest_sales:,.2f}")
print(f"Lowest Sale: ₹{lowest_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Number of Orders: {number_of_orders}")
```

## 11. Sales Analysis by Category

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Highest_Sale=("Total_Sales", "max"),
    Lowest_Sale=("Total_Sales", "min"),
    Quantity_Sold=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values("Total_Sales", ascending=False)

print(category_analysis)
```

## 12. Sales Analysis by City

```python
city_analysis = df.groupby("City").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values("Total_Sales", ascending=False)

print(city_analysis)
```

## 13. Sales Analysis by Product

```python
product_analysis = df.groupby("Product").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values("Total_Sales", ascending=False)

print("Products ranked by Total Sales:")
print(product_analysis)
```

## 14. Sales Analysis by Payment Method

```python
payment_analysis = df.groupby("Payment_Method").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values("Total_Sales", ascending=False)

print(payment_analysis)
```

## 15. Sales Analysis by Region

```python
region_analysis = df.groupby("Region").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Quantity_Sold=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values("Total_Sales", ascending=False)

print(region_analysis)
```

## 16. Top-Performing Products

```python
top_products = product_analysis.head(5)

print("Top 5 products by total sales:")
print(top_products)
```

## 17. Highest-Selling Category and City

```python
best_category = category_analysis["Total_Sales"].idxmax()
best_city = city_analysis["Total_Sales"].idxmax()
best_product = product_analysis["Total_Sales"].idxmax()
best_payment = payment_analysis["Total_Sales"].idxmax()
best_region = region_analysis["Total_Sales"].idxmax()

print("Highest-selling category:", best_category)
print("Highest-selling city:", best_city)
print("Highest-selling product:", best_product)
print("Payment method with highest total sales:", best_payment)
print("Highest-selling region:", best_region)
```

The e-commerce dataset was successfully analyzed using Pandas. Data selection and filtering were used to identify relevant orders, while sorting helped find the highest- and lowest-value transactions. `groupby()` and aggregation functions provided comparisons across categories, cities, products, regions, and payment methods.

The analysis shows which products and categories contribute most to revenue and highlights differences in sales performance across locations and payment methods. These findings can be used to support product promotion, inventory planning, and sales strategy decisions.
